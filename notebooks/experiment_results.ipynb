# CI Diagnosis Policy Experiment

Verified five-seed shared-case experiment: 120 cases per seed, identical pre-generated observation opportunities for P0/P2/P3, 70% commitment threshold.

```python
import pandas as pd
import matplotlib.pyplot as plt

results = pd.DataFrame({
    'Policy': ['P0', 'P2', 'P3'],
    'Accuracy': [75.83, 71.00, 73.17],
    'Mean diagnostic cost': [9.00, 5.83, 5.36],
    'Mean decision cost': [14.25, 10.75, 10.63],
    'Mean actions': [6.00, 4.28, 3.64],
    'Escalation rate': [65.67, 61.50, 65.83],
})
results
```

```python
ax = results.plot(x='Policy', y='Accuracy', kind='bar', legend=False)
ax.set_ylabel('Accuracy (%)')
ax.set_title('Diagnostic accuracy')
plt.tight_layout()
plt.show()
```

```python
ax = results.plot(x='Policy', y='Mean decision cost', kind='bar', legend=False)
ax.set_ylabel('Relative decision cost')
ax.set_title('Decision cost')
plt.tight_layout()
plt.show()
```

## Threshold sensitivity

P3 is not universally dominant. At 60% and 70% it has lower mean decision cost and higher accuracy than P2; at 80%, P2 has slightly higher accuracy while P3 remains cheaper.

```python
sensitivity = pd.DataFrame({
    'Threshold': [60, 60, 70, 70, 80, 80],
    'Policy': ['P2', 'P3', 'P2', 'P3', 'P2', 'P3'],
    'Accuracy': [66.17, 72.50, 71.00, 73.17, 75.17, 74.50],
    'Mean diagnostic cost': [4.53, 4.60, 5.83, 5.36, 6.75, 6.09],
    'Mean decision cost': [10.03, 9.86, 10.75, 10.63, 11.98, 11.51],
    'Mean actions': [3.63, 3.19, 4.28, 3.64, 4.88, 4.19],
})
sensitivity
```
