# CI Diagnosis Policy Experiment

This notebook documents the final shared-case experiment. P0, P2 and P3 are evaluated on identical simulated cases. The numerical results below are the current five-seed aggregate at a 70% commitment threshold.

```python
import pandas as pd
import matplotlib.pyplot as plt

results = pd.DataFrame({
    'Policy': ['P0', 'P2', 'P3'],
    'Accuracy': [73.50, 69.00, 71.83],
    'Mean diagnostic cost': [9.00, 5.86, 5.25],
    'Mean actions': [6.00, 4.30, 3.57],
    'Escalation rate': [5.33, 4.50, 4.00],
})
results
```

```python
ax = results.plot(x='Policy', y='Accuracy', kind='bar', legend=False)
ax.set_ylabel('Accuracy (%)')
ax.set_title('Diagnostic accuracy')
plt.tight_layout()
plt.show()
```

```python
ax = results.plot(x='Policy', y='Mean diagnostic cost', kind='bar', legend=False)
ax.set_ylabel('Relative diagnostic cost')
ax.set_title('Diagnostic cost')
plt.tight_layout()
plt.show()
```

## Threshold sensitivity

The current analysis uses 60%, 70% and 80% commitment thresholds. The key finding is that P3 is not universally dominant: its advantage is primarily an efficiency/accuracy trade-off rather than a guaranteed accuracy win.

```python
sensitivity = pd.DataFrame({
    'Threshold': [60, 60, 70, 70, 80, 80],
    'Policy': ['P2', 'P3', 'P2', 'P3', 'P2', 'P3'],
    'Accuracy': [66.17, 72.50, 71.00, 73.17, 75.17, 74.50],
    'Mean cost': [4.53, 4.60, 5.83, 5.36, 6.75, 6.09],
    'Mean actions': [3.63, 3.19, 4.28, 3.64, 4.88, 4.19],
})
sensitivity
```
